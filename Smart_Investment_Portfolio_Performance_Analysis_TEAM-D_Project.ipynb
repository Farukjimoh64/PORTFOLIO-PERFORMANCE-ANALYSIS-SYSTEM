## Questions 21–27 — Asset-level performance

### Question 21 — Basic profit or loss

Profit or loss is the difference between what an asset is worth now and what was originally invested in it.

For this first calculation, transaction costs and dividend income are deliberately excluded so the basic concept is easy to see.

```python
# Question 21 — BASIC PROFIT OR LOSS

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]
    profit_loss = current_value - initial_investment

    if profit_loss >= 0:
        status = "Profit"
    else:
        status = "Loss"

    print(asset["asset_name"], ":", round(profit_loss, 2), "->", status)
```

### Question 22 — Understanding the main profit/loss formula

The basic formula is:

**Profit/Loss = Current Value − Initial Investment**

The next question improves this by considering the extra money received or spent through the investment.

```python
# Question 22 — DISPLAY THE BASIC PROFIT/LOSS FORMULA FOR EACH ASSET

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]
    profit_loss = current_value - initial_investment

    print(
        asset["asset_name"],
        "| Initial:", round(initial_investment, 2),
        "| Current:", round(current_value, 2),
        "| P/L:", round(profit_loss, 2)
    )
```

### Question 23 — Adding the transaction costs and dividend income
 
Transaction cost is money spent on buying/selling the investment. Dividend income is money received from an investment.

So the complete formula is:

**Net Profit/Loss = Current Value − Initial Investment − Transaction Cost + Dividend Income**

```python
# Question 23 — NET PROFIT/LOSS INCLUDING COSTS AND DIVIDENDS

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    print(asset["asset_name"], ":", round(net_profit_loss, 2))
```

### Question 24 — Calculate the investment value

The **initial investment** tells us how much was put into an asset at purchase.

The **current value** tells us what the quantity is worth at the current price.

These two values are the foundation for the later return calculation; the Return Percentages among others.

```python
# Question 24 — INITIAL INVESTMENT AND CURRENT VALUE

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    print("\n", asset["asset_name"])
    print("Initial Investment:", round(initial_investment, 2))
    print("Current Value:", round(current_value, 2))
```

### Question 25 — Return percentage
  
Return Percentage tells us the size of the profit or loss compared with the original investment.

So, the formula is:

**Return Percentage (%) = (Net Profit/Loss / Initial Investment) × 100**

```python
# Question 25 — RETURN PERCENTAGE

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    if initial_investment != 0:
        return_percentage = (net_profit_loss / initial_investment) * 100
    else:
        return_percentage = 0

    print(asset["asset_name"], ":", round(return_percentage, 2), "%")
```

### Question 26 — Performance classification

To make the percentages easier for even a lay-man to read, returns are placed into simple categories.

The rule used based on the directive or instruction here is:

- **15% or more** -> Excellent
- **5% to below 15%** -> Positive
- **0% to below 5%** -> Low Positive
- **Below 0%** -> Loss

```python
# Question 26 — PERFORMANCE CLASSIFICATION

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]
    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    return_percentage = (net_profit_loss / initial_investment) * 100

    if return_percentage >= 15:
        performance = "Excellent"
    elif return_percentage >= 5:
        performance = "Positive"
    elif return_percentage >= 0:
        performance = "Low Positive"
    else:
        performance = "Loss"

    print(asset["asset_name"], ":", round(return_percentage, 2), "%", performance)
```

## Questions 28–32 — Portfolio-level performance

### Question 28 — Calculate portfolio totals

A portfolio contains many investments, so we need running totals. Since we are calculating one asset at a time, what happens to the entire portfolio?

The variables begin at zero because they act as **accumulators** and they are being updated. Each asset adds another amount to the running total.

That is, they are being initialized at zero, and updated over time at every iteration.

```python
# Question 28 — PORTFOLIO TOTALS

total_investment = 0
total_current_value = 0
total_profit = 0

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    total_investment += initial_investment
    total_current_value += current_value
    total_profit += net_profit_loss

print("Total Initial Investment:", f"{total_investment:,.2f}")
print("Total Current Value:     ", f"{total_current_value:,.2f}")
print("Total Net Profit/Loss:   ", f"{total_profit:,.2f}")
```

### Question 29 — Why initialize totals with zero?

Imagine counting money from several envelopes, or counting the stock; or taking inventory...

Before opening the first envelope, the running total is **0**. After opening each envelope, its value is added to the running total.

That is exactly what `+=` or `-=` does in the loop; Initialization and Update of the initial by an **accumulator**

```python
# Question 29 — SIMPLE ACCUMULATOR DEMONSTRATION

running_total = 0

for asset in portfolio:
    investment = asset["quantity"] * asset["purchase_price"]
    running_total += investment

print("Final running total:", f"{running_total:,.2f}")
```

### Question 30 — Portfolio totals plus performance breakdown

```python
# Question 30 — PORTFOLIO TOTALS AND PERFORMANCE COUNTS

total_investment = 0
total_current_value = 0
total_profit = 0

performance_counts = {
    "Excellent": 0,
    "Positive": 0,
    "Low Positive": 0,
    "Loss": 0
}

for asset in portfolio:
    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    return_percentage = (net_profit_loss / initial_investment) * 100

    if return_percentage >= 15:
        performance = "Excellent"
    elif return_percentage >= 5:
        performance = "Positive"
    elif return_percentage >= 0:
        performance = "Low Positive"
    else:
        performance = "Loss"

    performance_counts[performance] += 1
    total_investment += initial_investment
    total_current_value += current_value
    total_profit += net_profit_loss

portfolio_return = (total_profit / total_investment) * 100 if total_investment else 0

print("=" * 55)
print("PORTFOLIO TOTALS")
print("=" * 55)
print("Initial Investment:", f"{total_investment:,.2f}")
print("Current Value:    ", f"{total_current_value:,.2f}")
print("Net Profit/Loss:  ", f"{total_profit:,.2f}")
print("Overall Return:   ", f"{portfolio_return:.2f}%")
print("\nPerformance Breakdown:")
for label, count in performance_counts.items():
    print(f"{label:<15} {count}")
```

# INTEGRATED ANALYST SYSTEM
## From individual answers (manually) to one reusable analysis system

The next section turns the work into a cleaner **mini-system** with the help of established or created functions 

A function is a reusable block of instructions. Instead of rewriting the same formulas manually every time, the system can calculate them once and reuse the result.

```python
# ============================================================
# CORE ANALYSIS FUNCTIONS
# ============================================================

def calculate_asset_metrics(asset):
    """Calculate the main performance figures for one asset."""

    initial_investment = asset["quantity"] * asset["purchase_price"]
    current_value = asset["quantity"] * asset["current_price"]

    net_profit_loss = (
        current_value
        - initial_investment
        - asset["transaction_cost"]
        + asset["dividend_income"]
    )

    if initial_investment != 0:
        return_percentage = (net_profit_loss / initial_investment) * 100
    else:
        return_percentage = 0

    if return_percentage >= 15:
        performance = "Excellent"
    elif return_percentage >= 5:
        performance = "Positive"
    elif return_percentage >= 0:
        performance = "Low Positive"
    else:
        performance = "Loss"

    return {
        "initial_investment": initial_investment,
        "current_value": current_value,
        "net_profit_loss": net_profit_loss,
        "return_percentage": return_percentage,
        "performance": performance
    }


def calculate_portfolio_summary(portfolio):
    """Calculate overall portfolio figures."""

    total_investment = 0
    total_current_value = 0
    total_profit = 0

    performance_counts = {
        "Excellent": 0,
        "Positive": 0,
        "Low Positive": 0,
        "Loss": 0
    }

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)

        total_investment += metrics["initial_investment"]
        total_current_value += metrics["current_value"]
        total_profit += metrics["net_profit_loss"]
        performance_counts[metrics["performance"]] += 1

    if total_investment != 0:
        portfolio_return = (total_profit / total_investment) * 100
    else:
        portfolio_return = 0

    return {
        "total_investment": total_investment,
        "total_current_value": total_current_value,
        "total_profit": total_profit,
        "portfolio_return": portfolio_return,
        "performance_counts": performance_counts
    }
```

### Why these functions help

The function does the work and returns the answer.

This makes the final project cleaner while keeping the underlying calculations understandable.

```python
# ============================================================
# FULL PORTFOLIO REPORT
# ============================================================

def display_portfolio_report(portfolio):
    summary = calculate_portfolio_summary(portfolio)

    print("=" * 65)
    print("SMARTINVEST AFRICA — PORTFOLIO PERFORMANCE REPORT")
    print("=" * 65)

    print(f"Number of investments: {len(portfolio)}")
    print(f"Total initial investment: {summary['total_investment']:,.2f}")
    print(f"Total current value:      {summary['total_current_value']:,.2f}")
    print(f"Total net profit/loss:    {summary['total_profit']:,.2f}")
    print(f"Overall portfolio return: {summary['portfolio_return']:.2f}%")

    print("\nPerformance breakdown:")
    for category, count in summary["performance_counts"].items():
        print(f"  {category:<15}: {count}")

    print("\nIndividual asset performance:")
    print("-" * 65)

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)
        print(
            f"{asset['asset_name']:<25}"
            f" {metrics['return_percentage']:>7.2f}%"
            f"  {metrics['performance']}"
        )

display_portfolio_report(portfolio)
```

# RISK FLAGS AND PORTFOLIO HEALTH

It identifies conditions in the fictional dataset that deserve attention.

```python
# ============================================================
# RISK FLAGS
# ============================================================

def generate_risk_flags(portfolio):
    flags = []

    summary = calculate_portfolio_summary(portfolio)
    total_current_value = summary["total_current_value"]

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)
        weight = (metrics["current_value"] / total_current_value) * 100

        if metrics["return_percentage"] < 0:
            flags.append(
                asset["asset_name"] + " has a negative return."
            )

        if weight > 30:
            flags.append(
                asset["asset_name"] + " exceeds the 30% concentration threshold."
            )

    # Country concentration check
    country_values = {}

    for asset in portfolio:
        current_value = asset["quantity"] * asset["current_price"]
        country = asset["country"]

        if country not in country_values:
            country_values[country] = 0

        country_values[country] += current_value

    for country, value in country_values.items():
        country_weight = (value / total_current_value) * 100

        if country_weight > 50:
            flags.append(
                country + " represents more than 50% of portfolio value."
            )

    return flags


risk_flags = generate_risk_flags(portfolio)

print("RISK FLAGS")
print("=" * 55)

if risk_flags:
    for number, flag in enumerate(risk_flags, start=1):
        print(number, ".", flag)
else:
    print("No defined risk flags were triggered.")
```

## Portfolio Health Score

The system starts at 100 and subtracts points for:
- invalid data
- loss-making assets
- concentration flags
- and many other errors or limitations
- 
This makes the demonstration of how several checks can be combined into one summary indicator.

```python
# ============================================================
# PORTFOLIO HEALTH SCORE
# ============================================================

def calculate_health_score(portfolio):
    score = 100

    # Data-quality penalty
    invalid_fields = 0

    for asset in portfolio:
        if asset["quantity"] <= 0:
            invalid_fields += 1
        if asset["purchase_price"] <= 0:
            invalid_fields += 1
        if asset["current_price"] < 0:
            invalid_fields += 1

    score -= invalid_fields * 10

    # Loss-making asset penalty
    loss_count = 0

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)

        if metrics["net_profit_loss"] < 0:
            loss_count += 1

    score -= loss_count * 3

    # Concentration penalty
    summary = calculate_portfolio_summary(portfolio)

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)
        weight = (metrics["current_value"] / summary["total_current_value"]) * 100

        if weight > 30:
            score -= 10

    # Keep score between 0 and 100
    if score < 0:
        score = 0

    if score > 100:
        score = 100

    return score


health_score = calculate_health_score(portfolio)

print("Portfolio Health Score:", health_score, "/ 100")
```

# MANAGEMENT SUMMARY

A management summary converts calculations or qualitative reports into qualitative meanings that are understandable even to someone who is not an analyst

It should answer:
1. How much was invested?
2. What is the current value?
3. What is the net result?
4. How many assets are profitable/loss-making?
5. Which assets had the highest and lowest returns?
6. Were any defined risk flags triggered?

These have been discussed above and the calculated in the initial records

The observations below are generated from the dataset rather than manually typed, the reason for created functions

```python
# ============================================================
# MANAGEMENT SUMMARY
# ============================================================

def get_best_and_worst(portfolio):
    best_asset = None
    worst_asset = None
    best_return = None
    worst_return = None

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)
        current_return = metrics["return_percentage"]

        if best_return is None or current_return > best_return:
            best_return = current_return
            best_asset = asset["asset_name"]

        if worst_return is None or current_return < worst_return:
            worst_return = current_return
            worst_asset = asset["asset_name"]

    return best_asset, best_return, worst_asset, worst_return


def display_management_summary(portfolio):
    summary = calculate_portfolio_summary(portfolio)
    best_asset, best_return, worst_asset, worst_return = get_best_and_worst(portfolio)
    flags = generate_risk_flags(portfolio)
    health_score = calculate_health_score(portfolio)

    profitable = 0
    loss_making = 0

    for asset in portfolio:
        metrics = calculate_asset_metrics(asset)

        if metrics["net_profit_loss"] >= 0:
            profitable += 1
        else:
            loss_making += 1

    print("=" * 70)
    print("SMARTINVEST AFRICA — MANAGEMENT SUMMARY")
    print("=" * 70)

    print(f"Investments analysed: {len(portfolio)}")
    print(f"Total initial investment: {summary['total_investment']:,.2f}")
    print(f"Current portfolio value:  {summary['total_current_value']:,.2f}")
    print(f"Net portfolio result:    {summary['total_profit']:,.2f}")
    print(f"Overall portfolio return: {summary['portfolio_return']:.2f}%")

    print("=" * 70)
    print("PORTFOLIO SUMMARY")
    print("=" * 70)
    print("Number of investments:", len(portfolio))
    print("Total Initial Investment:", f"{total_investment:,.2f}")
    print("Total Current Value:     ", f"{total_current_value:,.2f}")
    print("Total Net Profit/Loss:   ", f"{total_profit:,.2f}")

    print("=" * 70)
    print("PERFORMANCE:")
    print("=" * 70)
    print(f"  Profitable assets: {profitable}")
    print(f"  Loss-making assets: {loss_making}")

    print("=" * 70)
    print("Highest return:")
    print(f"  {best_asset} — {best_return:.2f}%")

    print("=" * 70)
    print("Lowest return:")
    print(f"  {worst_asset} — {worst_return:.2f}%")

    print("=" * 70)
    print("Defined risk flags:")
    print("=" * 70)
    if flags:
        for flag in flags:
            print("  -", flag)
    else:
        print("  None triggered.")

    print("=" * 70)
    print(f"Portfolio health score: {health_score}/100")

    print("=" * 70)
    print("Analyst note:")
    print(
        "The figures above describe the fictional dataset. "
        "They are not investment recommendations."
    )
    print("=" * 70)
display_management_summary(portfolio)
```
